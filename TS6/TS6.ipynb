## **Introducción:**
En el procesamiento analógico de señales, los filtros pasivos constituidos por resistencias ($R$), inductores ($L$) y capacitores ($C$) representan las estructuras fundamentales para seleccionar, atenuar o moldear bandas de frecuencia específicas antes de las etapas de digitalización o amplificación. El estudio de los sistemas lineales e invariantes en el tiempo (LTI) de segundo orden resulta central en la teoría de circuitos, ya que estos sistemas introducen pares de polos complejos conjugados capaces de moldear la respuesta en frecuencia con diferentes grados de amortiguamiento y selectividad.

<br>
<br>

**Función de Transferencia: Polos y Ceros del Sistema**
<br>
Un filtro analógico LTI en tiempo continuo se caracteriza en el dominio de Laplace ($s = \sigma + j\omega$) mediante su función de transferencia $H(s)$, expresada como el cociente de dos polinomios: $$H(s) = \frac{V_2(s)}{V_1(s)} = \frac{N(s)}{D(s)} = K \cdot \frac{(s - z_1)(s - z_2) \cdots (s - z_m)}{(s - p_1)(s - p_2) \cdots (s - p_n)}$$
<br>
- *Polos ($p_i$):* Son las raíces del denominador ($D(s) = 0$). Determinan la estabilidad BIBO del sistema (que exige $\text{Re}(p_i) < 0$) y la dinámica temporal de los transitorios .
<br>
- *Ceros ($z_i$):* Son las raíces del numerador ($N(s) = 0$). Se obtienen igualando el polinomio del numerador a cero ($N(s) = 0$) y representan las frecuencias complejas en las cuales la transmisión del sistema se anula por completo ($|H(z_i)| = 0$).
<br>
- *Rol de los Ceros en la Selección de Banda:* Mientras que los polos dictan dónde el sistema responde o resuena, la cantidad y ubicación de los ceros (en el origen $s=0$, finitos en el plano $s$ o en el infinito $s \to \infty$) establecen el comportamiento del filtro a bajas y altas frecuencias, definiendo si se trata de un filtro Pasa-Bajos, Pasa-Altos o Pasa-Banda. 

<br>
<br>

**Formas Canónicas de Segundo Orden y Factor de Calidad ($Q$)**
<br>
Las funciones de transferencia de segundo orden sin ceros en el semiplano derecho se parametrizan en función de dos variables fundamentales: la frecuencia natural de resonancia ($\omega_0$) y el factor de calidad ($Q$):
<br>
- *Filtro Pasa-Bajos (LPF):* $$H_{LP}(s) = \frac{\omega_0^2}{s^2 + \frac{\omega_0}{Q} s + \omega_0^2}$$
<br>
- *Filtro Pasa-Altos (HPF):* $$H_{HP}(s) = \frac{s^2 + s \frac{\omega_0}{Q}}{s^2 + s \frac{\omega_0}{Q} + \omega_0^2}$$
<br>
- *Filtro Pasa-Banda (BPF):* $$H_{BP}(s) = \frac{s \frac{\omega_0}{Q}}{s^2 + s \frac{\omega_0}{Q} + \omega_0^2}$$
<br>
- *Significado del Factor de Calidad ($Q$):*
<br>
  - Para $Q < 0{,}5$: El sistema es subamortiguado/sobreamortiguado con polos reales distintos.
<br>
  - Para $Q = 0{,}5$: Polos reales dobles (amortiguamiento crítico).
<br>
  - Para $Q > \frac{1}{\sqrt{2}} \approx 0{,}707$: Los polos forman un par complejo conjugado cercano al eje imaginario, lo que genera una sobreelevación o resalto de resonancia (peak) en la magnitud de la respuesta en frecuencia cerca de $\omega_0$

<br>
<br>

**Respuesta de Fase y Denominación del Comportamiento del Circuito**
<br>
La respuesta de fase $\phi(\omega) = \arg(H(j\omega))$ refleja la relación temporal entre las señales de salida e entrada, clasificándose según el signo del ángulo de desfase:
<br>
- Carácter Capacitivo (Retraso de Fase, $\phi < 0$): La señal de salida retrasa respecto de la entrada. Típico de filtros pasa-bajos o regiones donde predomina la reactancia del capacitor.
<br>
- Carácter Inductivo (Adelanto de Fase, $\phi > 0$): La señal de salida adelanta respecto de la entrada. Aparece en filtros pasa-altos o zonas dominadas por la presencia del inductor.
<br>
- Carácter Resistivo (En Fase, $\phi = 0^\circ$): La salida y la entrada están perfectamente sincronizadas. Ocurre en la frecuencia de resonancia/antirresonancia de los filtros pasa-banda, donde los efectos reactivos de $L$ y $C$ se cancelan mutuamente.

<br>
<br>

**Respuesta Asintótica y Diagramas de Bode**
<br>
El comportamiento frecuencial se representa mediante Diagramas de Bode:
- *Módulo:* Expresado en escala logarítmica como $20 \log_{10} |H(j\omega)|$ [dB].
- *Fase:* Expresada en grados $\phi(\omega) = \arg(H(j\omega))$.

<br>

Cada polo o cero de segundo orden aporta una pendiente asintótica de $\pm 40\text{ dB/década}$ ($\pm 12\text{ dB/octava}$) en las frecuencias alejadas de $\omega_0$. El Ancho de Banda ($B$) se define como el rango de frecuencias donde la potencia de salida no cae más de $3\text{ dB}$ respecto del valor nominal en la banda pasante.

<br>
<br>

## **Desarrollo**

### **Circuito A**

**Obtención Analítica de la Función de Transferencia $H(s)$:**
<br>
Considerando la salida en circuito abierto ($I_2 = 0$), aplicamos el divisor de tensión en el dominio de Laplace ($s$):$$H(s) = \frac{V_2(s)}{V_1(s)} = \frac{Z_C(s)}{Z_L(s) + Z_R(s) + Z_C(s)}$$

Sustituyendo las impedancias $Z_L(s) = sL$, $Z_R(s) = R$ y $Z_C(s) = \frac{1}{sC}$:$$H(s) = \frac{\frac{1}{sC}}{sL + R + \frac{1}{sC}} = \frac{1}{s^2 LC + sRC + 1} = \frac{\frac{1}{LC}}{s^2 + s \frac{R}{L} + \frac{1}{LC}}$$

Forma Canónica de Segundo Orden: La expresión estándar de un filtro pasa-bajos de segundo orden es: $$H(s) = \frac{\omega_0^2}{s^2 + \frac{\omega_0}{Q} s + \omega_0^2}$$
Donde:
- Frecuencia natural de resonancia: $\omega_0 = \frac{1}{\sqrt{LC}}$
- Factor de Calidad: $Q = \frac{\omega_0 L}{R} = \frac{1}{R} \sqrt{\frac{L}{C}}$

Sustitución de los datos: Con $\omega_0 = 1\text{ rad/s}$, $R = 1\ \Omega$ y $Q = 2\sqrt{2}$:
- De $Q = \frac{\omega_0 L}{R} \implies 2\sqrt{2} = \frac{1 \cdot L}{1} \implies \mathbf{L = 2\sqrt{2}\ \text{H}}$
- De $\omega_0^2 = \frac{1}{LC} = 1 \implies \mathbf{C = \frac{1}{2\sqrt{2}}\ \text{F} = \frac{\sqrt{2}}{4}\ \text{F}}$
- Coeficiente del término lineal: $\frac{\omega_0}{Q} = \frac{1}{2\sqrt{2}} = \frac{\sqrt{2}}{4}$

Por lo tanto, la función de transferencia analítica final es:$$\mathbf{H(s) = \frac{1}{s^2 + \frac{\sqrt{2}}{4} s + 1} = \frac{1}{s^2 + \frac{1}{2\sqrt{2}} s + 1}}$$

**Respuesta de Módulo y Fase** 
<br>
Evaluando en el eje de frecuencias reales $s = j\omega$:$$H(j\omega) = \frac{1}{(1 - \omega^2) + j \omega \frac{\sqrt{2}}{4}}$$

*Módulo:*
$|H(j\omega)|$:$$|H(j\omega)| = \frac{1}{\sqrt{(1 - \omega^2)^2 + \frac{\omega^2}{8}}}$$

- Respuesta Asintótica:
  - Para $\omega \ll \omega_0 = 1\text{ rad/s}$: $|H(j\omega)| \to 1 \quad (\mathbf{0\text{ dB}})$.
  - Para $\omega \gg \omega_0 = 1\text{ rad/s}$: $|H(j\omega)| \approx \frac{1}{\omega^2}$, con una pendiente asintótica de $-40\text{ dB/década}$ ($-12\text{ dB/octava}$).
<br>
<br>
- Comportamiento en Resonancia ($\omega = \omega_0 = 1$):
  - $|H(j1)| = \frac{1}{\sqrt{1/8}} = 2\sqrt{2} \approx 2.83 \quad (\mathbf{+9.03\text{ dB}})$.
  - Dado que $Q = 2\sqrt{2} > \frac{1}{\sqrt{2}}$, presenta un pico de sobreelevación (resalto espectral) en la frecuencia de sobreimpulso $\omega_p = \sqrt{1 - \frac{1}{2Q^2}} = \sqrt{\frac{15}{16}} \approx 0.97\text{ rad/s}$.

<br>

*Fase $\phi(\omega)$:*
$$\phi(\omega) = -\arctan\left( \frac{\frac{\sqrt{2}}{4} \omega}{1 - \omega^2} \right)$$
- Respuesta Asintótica:
  - Para $\omega \le 0.1\text{ rad/s}$: $\phi(\omega) \to \mathbf{0^\circ}$.
  - Para $\omega = \omega_0 = 1\text{ rad/s}$: $\phi(1) = \mathbf{-90^\circ}$.
  - Para $\omega \ge 10\text{ rad/s}$: $\phi(\omega) \to \mathbf{-180^\circ}$.

**Respuestas a las Preguntas de la Consigna**
<br>
<br>
*¿De qué tipo de filtro se trata?* Es un Filtro Pasa-Bajos de segundo orden (con sobreelevación o resalto de resonancia debido a que $Q > \frac{1}{\sqrt{2}}$). Deja pasar sin atenuar las componentes continuas y de baja frecuencia ($\omega < 1\text{ rad/s}$) y atenúa drásticamente las frecuencias altas a razón de $-40\text{ dB/década}$.
<br>
<br>
*¿Cuál es el ancho de banda del filtro?* El ancho de banda $B$ (o $\omega_c$) en un filtro pasa-bajos se define como la frecuencia de corte a $-3\text{ dB}$, donde la potencia cae a la mitad ($|H(j\omega_c)|^2 = \frac{|H(0)|^2}{2} = \frac{1}{2}$):$$(1 - \omega_c^2)^2 + \frac{\omega_c^2}{8} = 2 \implies \omega_c^4 - \frac{15}{8} \omega_c^2 - 1 = 0$$Resolviendo la ecuación cuadrática para $\omega_c^2$:$$\omega_c^2 = \frac{15 + \sqrt{481}}{16} \approx 2.308 \implies \mathbf{B = \omega_c \approx 1.52\text{ rad/s}} \quad \left(f_c = \frac{\omega_c}{2\pi} \approx 0.242\text{ Hz}\right)$$

### **Circuito B**

<br>

**Obtención Analítica de la Función de Transferencia $H(s)$**
<br>
Trabajando en el dominio de Laplace ($s$) en circuito abierto ($I_2 = 0$):
<br>
- Impedancia del bloque serie $R \parallel C$: $$Z_{RC}(s) = \frac{R \cdot \frac{1}{sC}}{R + \frac{1}{sC}} = \frac{R}{1 + sRC}$$
<br>
- Impedancia de la rama en derivación (Inductor $L$): $$Z_L(s) = sL$$
<br>
- Divisor de Tensión para $V_2(s)$: $$H(s) = \frac{V_2(s)}{V_1(s)} = \frac{Z_L(s)}{Z_{RC}(s) + Z_L(s)} = \frac{sL}{\frac{R}{1 + sRC} + sL}$$
<br>

Multiplicando numerador y denominador por $(1 + sRC)$: $$H(s) = \frac{sL(1 + sRC)}{R + sL(1 + sRC)} = \frac{s^2 RLC + sL}{s^2 RLC + sL + R}$$
<br>
Dividiendo todo por $RLC$: $$\mathbf{H(s) = \frac{s^2 + s \frac{1}{RC}}{s^2 + s \frac{1}{RC} + \frac{1}{LC}}}$$

<br>
<br>

**Parametrización y Valores de Componentes**
<br>
Comparando el denominador con la forma canónica de 2do orden $s^2 + s \frac{\omega_0}{Q} + \omega_0^2$:
- Frecuencia natural: $\omega_0 = \frac{1}{\sqrt{LC}}$
- Coeficiente del término lineal: $\frac{\omega_0}{Q} = \frac{1}{RC} \implies \mathbf{Q = \omega_0 R C = R \sqrt{\frac{C}{L}}}$
<br>

Reescribiendo la transferencia en función de $\omega_0$ y $Q$: $$H(s) = \frac{s^2 + s \frac{\omega_0}{Q}}{s^2 + s \frac{\omega_0}{Q} + \omega_0^2}$$
<br>
Sustitución con los datos($\omega_0 = 1\text{ rad/s}$, $R = 1\ \Omega$, $Q = 2\sqrt{2}$):
- De $Q = \omega_0 R C \implies 2\sqrt{2} = 1 \cdot 1 \cdot C \implies \mathbf{C = 2\sqrt{2}\ \text{F} \approx 2.828\ \text{F}}$
<br>
- De $\omega_0^2 = \frac{1}{LC} = 1 \implies L = \frac{1}{C} = \frac{1}{2\sqrt{2}} \implies \mathbf{L = \frac{\sqrt{2}}{4}\ \text{H} \approx 0.3536\ \text{H}}$
<br>

La función de transferencia numérica resultante es: $$\mathbf{H(s) = \frac{s^2 + \frac{\sqrt{2}}{4} s}{s^2 + \frac{\sqrt{2}}{4} s + 1}}$$

<br>
<br

**Respuesta de Frecuencia**
<br>
Sustituyendo $s = j\omega$:
- Bajas Frecuencias ($\omega \ll 1\text{ rad/s}$): $H(j\omega) \approx j\omega \frac{L}{R} = j\omega \frac{\sqrt{2}}{4}$.
  - Módulo: Crece a razón de $+20\text{ dB/década}$.
  - Fase: Tiende a $+90^\circ$.
<br>
<br>
- Altas Frecuencias ($\omega \gg 1\text{ rad/s}$): $H(j\omega) \to 1 \quad (\mathbf{0\text{ dB}})$.
  - Fase: Tiende a $0^\circ$.
<br>
<br>
- Frecuencia de Resonancia ($\omega = \omega_0 = 1\text{ rad/s}$): $H(j1) = 1 + jQ = 1 + j 2\sqrt{2}$.
  - Módulo en resonancia: $|H(j1)| = \sqrt{1 + Q^2} = \sqrt{1 + 8} = \mathbf{3} \quad (\approx \mathbf{+9.54\text{ dB}})$.
  - Fase en resonancia: $\phi(1) = \arctan(2\sqrt{2}) \approx \mathbf{70.53^\circ}$.

<br>
<br>

**Respuestas a las Preguntas de la Consigna**
<br>
<br>
*¿De qué tipo de filtro se trata?* Es un Filtro Pasa-Altos de segundo orden (posee un cero en el origen $s=0$, un cero finito en $s = -\frac{\omega_0}{Q}$ y un par de polos complejos conjugados con resalto de resonancia por ser $Q = 2\sqrt{2} > \frac{1}{\sqrt{2}}$). Deja pasar las altas frecuencias sin atenuación ($0\text{ dB}$) y atenúa las bajas frecuencias.
<br>
<br>
*¿Cuál es el ancho de banda del filtro?* El ancho de banda abarca desde la frecuencia de corte a $-3\text{ dB}$ (respecto a la banda pasante de $0\text{ dB}$) hasta infinito. Igualando $|H(j\omega_c)|^2 = \frac{1}{2}$: $$\omega_c^4 + \frac{17}{8}\omega_c^2 - 1 = 0 \implies \omega_c^2 = \frac{-\frac{17}{8} + \sqrt{\left(\frac{17}{8}\right)^2 + 4}}{2} \approx 0{,}3966 \implies \mathbf{\omega_c \approx 0{,}63\text{ rad/s}}$$Por lo tanto, la banda pasante del filtro es $\mathbf{B = [\omega_c, \infty) \approx [0.63, \infty)\text{ rad/s}}$.

<br>
<br>

### **Circuito C**

**Obtención Analítica de la Función de Transferencia $H(s)$**
<br>
Trabajando en el dominio de Laplace ($s$) en circuito abierto ($I_2 = 0$):
<br>
- Impedancia en paralelo $L \parallel C$: $$Z_p(s) = \frac{sL \cdot \frac{1}{sC}}{sL + \frac{1}{sC}} = \frac{sL}{s^2 LC + 1}$$
<br>
- Divisor de Tensión: $$H(s) = \frac{V_2(s)}{V_1(s)} = \frac{Z_p(s)}{R + Z_p(s)} = \frac{\frac{sL}{s^2 LC + 1}}{R + \frac{sL}{s^2 LC + 1}} = \frac{sL}{s^2 RLC + sL + R}$$
<br>
- Forma Canónica Dividiendo por $RLC$: $$\mathbf{H(s) = \frac{s \frac{1}{RC}}{s^2 + s \frac{1}{RC} + \frac{1}{LC}}}$$

<br>
<br>

**Parametrización y Valores Numéricos**
<br>
Comparando con la forma canónica de un Filtro Pasa-Banda de 2º orden:$$H(s) = \frac{s \frac{\omega_0}{Q}}{s^2 + s \frac{\omega_0}{Q} + \omega_0^2}$$
<br>
- Frecuencia central: $\omega_0 = \frac{1}{\sqrt{LC}}$
- Ancho de banda / término lineal: $\frac{\omega_0}{Q} = \frac{1}{RC} \implies \mathbf{Q = \omega_0 R C = R \sqrt{\frac{C}{L}}}$

<br>

Sustitución con los datos del enunciado ($\omega_0 = 1\text{ rad/s}$, $R = 1\ \Omega$, $Q = 2\sqrt{2}$):
<br>
- De $Q = \omega_0 R C \implies 2\sqrt{2} = 1 \cdot 1 \cdot C \implies \mathbf{C = 2\sqrt{2}\ \text{F} \approx 2.8284\ \text{F}}$
<br>
- De $\omega_0^2 = \frac{1}{LC} = 1 \implies L = \frac{1}{C} = \frac{1}{2\sqrt{2}} \implies \mathbf{L = \frac{\sqrt{2}}{4}\ \text{H} \approx 0.3536\ \text{H}}$
<br>
- Término del numerador y lineal: $\frac{\omega_0}{Q} = \frac{1}{2\sqrt{2}} = \frac{\sqrt{2}}{4}$

<br>

La función de transferencia analítica reemplazada es:$$\mathbf{H(s) = \frac{s \frac{\sqrt{2}}{4}}{s^2 + s \frac{\sqrt{2}}{4} + 1} = \frac{0.3536 s}{s^2 + 0.3536 s + 1}}$$

<br>
<br>

**Respuesta de Frecuencia**
<br>
Evaluando en el eje $s = j\omega$:$$H(j\omega) = \frac{j \omega \frac{\sqrt{2}}{4}}{(1 - \omega^2) + j \omega \frac{\sqrt{2}}{4}}$$
<br>
- Bajas Frecuencias ($\omega \ll 1\text{ rad/s}$):
  - $H(j\omega) \approx j \omega \frac{\sqrt{2}}{4}$.
  - Módulo: Crece con pendiente asintótica de $+20\text{ dB/década}$.
  - Fase: Tiende a $+90^\circ$.
<br>
<br>
- Altas Frecuencias ($\omega \gg 1\text{ rad/s}$):
  - $H(j\omega) \approx \frac{j \omega \frac{\sqrt{2}}{4}}{-\omega^2} = \frac{\sqrt{2}}{4 j \omega}$.
  - Módulo: Decae con pendiente asintótica de $-20\text{ dB/década}$.
  - Fase: Tiende a $-90^\circ$.
<br>
<br>
- Frecuencia Central de Resonancia ($\omega = \omega_0 = 1\text{ rad/s}$):
  - $H(j1) = \frac{j \frac{\sqrt{2}}{4}}{j \frac{\sqrt{2}}{4}} = 1$.
  - Módulo: $|H(j1)| = 1 \quad (\mathbf{0\text{ dB}})$.
  - Fase: $\phi(1) = \mathbf{0^\circ}$.

<br>
<br>

**Respuestas a las Preguntas de la Consigna**
<br>
<br>
*¿De qué tipo de filtro se trata?* Es un Filtro Pasa-Banda (BPF - Band-Pass Filter) de segundo orden. Posee un cero en el origen ($s=0$) y un par de polos complejos conjugados, lo que atenuará tanto las componentes de baja frecuencia como las de alta frecuencia, dejando pasar sin atenuación ($0\text{ dB}$) únicamente la banda centrada en $\omega_0 = 1\text{ rad/s}$.
<br>
<br>
*¿Cuál es el ancho de banda del filtro?* En un filtro pasa-banda canónico de segundo orden, el ancho de banda $B$ a $-3\text{ dB}$ está dado directamente por la relación:$$\mathbf{B = \frac{\omega_0}{Q} = \frac{1}{2\sqrt{2}} = \frac{\sqrt{2}}{4} \approx 0.3536\text{ rad/s}} \quad \left(f_B = \frac{B}{2\pi} \approx 0.0563\text{ Hz}\right)$$Frecuencias de corte a $-3\text{ dB}$:$\omega_{c1} \approx 0.839\text{ rad/s}$$\omega_{c2} \approx 1.192\text{ rad/s}$$B = \omega_{c2} - \omega_{c1} \approx 0.3536\text{ rad/s}$

## **Analisis de Resultados**

**Polos en Común y Estabilidad**
<br>
Al observar las tres funciones de transferencia obtenidas, las tres compartieron exactamente el mismo denominador $D(s) = s^2 + \frac{\sqrt{2}}{4} s + 1 = s^2 + 0{,}3536 s + 1$.
- *Origen:* El denominador de una función de transferencia determina la ecuación característica de la red RLC y depende únicamente de la malla/nodo donde se disipa la energía. Dado que la frecuencia natural de resonancia se fijó en $\omega_0 = 1\text{ rad/s}$ y el factor de calidad en $Q = 2\sqrt{2}$, las tres topologías presentan los mismos polos en el plano $s$: $$s_{1,2} = -0{,}1768 \pm j 0{,}9842$$
<br>
- *Implicación en la Respuesta Temporal:* Al estar ubicados estrictamente en el semiplano izquierdo ($\text{Re}(s) < 0$), los tres sistemas son BIBO estables. Su parte real ($\sigma = -0{,}1768$) impone la misma tasa de decaimiento exponencial frente a transitorios, mientras que la parte imaginaria ($\omega_d = 0{,}9842\text{ rad/s}$) define la frecuencia de oscilación amortiguada del sistema.

<br>
<br>

**El Rol de los Ceros** 
<br>
Si los tres filtros comparten los mismos polos, ¿qué determina que uno sea Pasa-Bajos, otro Pasa-Altos y el otro Pasa-Banda? La respuesta esta en la ubicación de los ceros en el numerador (definida por cuál componente se selecciona para tomar la tensión de salida $V_2$):
<br>
- *Filtro Pasa-Bajos (Inciso a - Salida sobre $C$):* No posee ceros finitos en el plano $s$. En continua ($s = 0$), el capacitor se comporta como un circuito abierto y la inductancia como un cortocircuito, transfiriendo la totalidad de la tensión de entrada a la salida ($|H(0)| = 1 \implies 0\text{ dB}$). A frecuencias muy altas ($s \to \infty$), la reactancia del capacitor tiende a cero, cortocircuitando la salida a masa con una pendiente de atenuación de $-40\text{ dB/década}$.
<br>
<br>
- *Filtro Pasa-Altos (Inciso b - Salida sobre $L$):* Introduce dos ceros en el numerador: uno en el origen ($s = 0$) y otro real en $s = -0{,}3536$. El cero en el origen bloquea por completo la componente continua ($DC$). A frecuencias muy altas, el bloque serie $R \parallel C$ pasa a comportarse como un cortocircuito (dominado por la baja impedancia del capacitor), transfiriendo toda la señal a la salida sobre el inductor ($|H(j\infty)| = 1 \implies 0\text{ dB}$).
<br>
<br>
- *Filtro Pasa-Banda (Inciso c - Salida sobre $L \parallel C$):* Presenta un único cero en el origen ($s = 0$). A bajas frecuencias, la inductancia bloquea la continua derivándola a masa; a altas frecuencias, el capacitor cortocircuita la salida a masa. La transmisión solo es máxima en la zona intermedia ($\omega \approx \omega_0 = 1\text{ rad/s}$), donde $L \parallel C$ entra en resonancia de paralela (antirresonancia), presentando una impedancia equivalente infinita que no carga a la resistencia serie.


**Filtros LPF y HPF:**
<br>
Dado que el factor de calidad es elevado ($Q = 2\sqrt{2} \approx 2{,}83$), supera el límite de $Q > 1/\sqrt{2} \approx 0{,}707$. Como resultado, tanto el pasa-bajos como el pasa-altos no presentan una atenuación progresiva suave, sino un pico de resonancia pronunciado de cerca de $+9\text{ dB}$ alrededor de $\omega_0 = 1\text{ rad/s}$. Esto refleja un sistema subamortiguado, donde la energía oscila intensamente entre el inductor y el capacitor antes de disiparse en la resistencia.

<br>

**Filtro BPF:**
<br>
En el pasa-banda, la ganancia máxima en la frecuencia de resonancia se mantiene acotada a $0\text{ dB}$ ($|H(j1)| = 1$). En este filtro, un $Q$ elevado no produce amplificación por encima de la señal de entrada, sino que se manifiesta como una alta selectividad: la banda de paso se vuelve muy angosta y concentrada, reduciendo el ancho de banda efectivo a solo $B = \frac{\omega_0}{Q} \approx 0{,}3536\text{ rad/s}$.

## **Aprendizajes**

<br>

- *Los ceros determinan el tipo de filtro:* Comprobamos que la ubicación de los polos en el plano $s$ define la estabilidad y la respuesta temporal del sistema, pero son los ceros (determinados por el componente sobre el que medimos la salida $V_2$) los que dictan si el circuito actúa como Pasa-Bajos, Pasa-Altos o Pasa-Banda.
<br>
<br>
- *Efecto de $Q > 0{,}707$:* Verificamos que un factor de calidad alto ($Q = 2\sqrt{2} \approx 2{,}83$) genera un resalto espectral de $\approx +9\text{ dB}$ en LPF y HPF por subamortiguamiento, mientras que en el BPF se traduce en una alta selectividad con un ancho de banda estrecho ($B = \omega_0 / Q$).
<br>
<br>
- *Relación entre topología y valores de componentes:* Comprobamos que para fijar los mismos valores de $Q$ y $\omega_0$, la forma en que se conecta la resistencia altera los valores necesarios de $L$ y $C$: cuando $R$ está en serie con el inductor exige una inductancia grande, mientras que al interactuar en paralelo con el capacitor pasa a exigir una capacitancia mayor.
<br>
<br>
- *Respuesta de fase en resonancia ($\omega_0 = 1\text{ rad/s}$):* Identificamos que en la frecuencia natural, el filtro Pasa-Banda presenta un desfase de $0^\circ$ ($L \parallel C$ entra en antirresonancia y el circuito se comporta como puramente resistivo), a diferencia del Pasa-Bajos que tiene caracter capacitivo ($-90^\circ$) y del Pasa-Altos, con caracter inductivo ($+70{,}5^\circ$).
