## **Introducción**
En el procesamiento analógico de señales, los filtros pasivos constituidos por resistencias ($R$), inductores ($L$) y capacitores ($C$) representan las estructuras fundamentales para seleccionar, atenuar o moldear bandas de frecuencia específicas antes de las etapas de digitalización o amplificación. El estudio de los sistemas lineales e invariantes en el tiempo (LTI) de segundo orden resulta central en la teoría de circuitos, ya que estos sistemas introducen pares de polos complejos conjugados capaces de moldear la respuesta en frecuencia con diferentes grados de amortiguamiento y selectividad.

**Función de Transferencia: Polos y Ceros del Sistema**

Un filtro analógico LTI en tiempo continuo se caracteriza en el dominio de Laplace ($s = \sigma + j\omega$) mediante su función de transferencia $H(s)$, expresada como el cociente de dos polinomios: $$H(s) = \frac{V_2(s)}{V_1(s)} = \frac{N(s)}{D(s)} = K \cdot \frac{(s - z_1)(s - z_2) \cdots (s - z_m)}{(s - p_1)(s - p_2) \cdots (s - p_n)}$$

- *Polos ($p_i$):* Son las raíces del denominador ($D(s) = 0$). Determinan la estabilidad BIBO del sistema (que exige $\text{Re}(p_i) < 0$) y la dinámica temporal de los transitorios.
- *Ceros ($z_i$):* Son las raíces del numerador ($N(s) = 0$). Representan las frecuencias complejas en las cuales la transmisión del sistema se anula por completo ($|H(z_i)| = 0$).
- *Rol de los Ceros en la Selección de Banda:* Mientras que los polos dictan dónde el sistema responde, la cantidad y ubicación de los ceros (en el origen $s=0$, finitos o en el infinito $s \to \infty$) establecen el comportamiento del filtro a bajas y altas frecuencias, definiendo si se trata de un filtro Pasa-Bajos, Pasa-Altos o Pasa-Banda.

**Formas Canónicas de Segundo Orden y Factor de Calidad ($Q$)**

Las funciones de transferencia de segundo orden se parametrizan en función de la frecuencia natural ($\omega_0$) y el factor de calidad ($Q$):
- *Filtro Pasa-Bajos (LPF):* $$H_{LP}(s) = \frac{\omega_0^2}{s^2 + \frac{\omega_0}{Q} s + \omega_0^2}$$
- *Filtro Pasa-Altos (HPF):* $$H_{HP}(s) = \frac{s^2 + s \frac{\omega_0}{Q}}{s^2 + s \frac{\omega_0}{Q} + \omega_0^2}$$
- *Filtro Pasa-Banda (BPF):* $$H_{BP}(s) = \frac{s \frac{\omega_0}{Q}}{s^2 + s \frac{\omega_0}{Q} + \omega_0^2}$$

**Significado del Factor de Calidad ($Q$):**
- Para $Q < 0.5$: Polos reales distintos (sobreamortiguado).
- Para $Q = 0.5$: Polos reales dobles en $-\omega_0$ (amortiguamiento crítico).
- Para $\mathbf{Q = \frac{1}{\sqrt{2}} \approx 0.707}$ (caso de diseño adoptado en este trabajo): Los polos forman un par complejo conjugado a $45^\circ$ del eje negativo real. Representa la aproximación de Butterworth de segundo orden (máximamente plana), que ofrece la mayor planicidad posible en la banda pasante sin presentar ninguna sobreelevación o pico de resonancia en el pasa-bajos.
- Para $Q > \frac{1}{\sqrt{2}}$: Los polos se acercan más al eje imaginario y provocan un sobreimpulso o resalto resonante (peak) en la magnitud cerca de $\omega_0$.

**Respuesta Asintótica y Diagramas de Bode**

El comportamiento frecuencial se representa mediante Diagramas de Bode:
- *Módulo:* $20 \log_{10} |H(j\omega)|$ [dB] vs. $\omega$ (escala logarítmica).
- *Fase:* $\phi(\omega) = \arg(H(j\omega))$ [grados] vs. $\omega$.

Cada polo o cero de segundo orden aporta una pendiente asintótica de $\pm 40\text{ dB/década}$ ($\pm 12\text{ dB/octava}$) alejada de $\omega_0$. El Ancho de Banda ($B$) se define como el rango de frecuencias donde la potencia de salida cae a la mitad ($-3\text{ dB}$) respecto del valor nominal en la banda pasante.

### **Circuito A — Filtro Pasa-Bajos (LPF)**

**Obtención Analítica de la Función de Transferencia $H(s)$:**
Considerando la salida en circuito abierto ($I_2 = 0$), aplicamos el divisor de tensión en el capacitor:
$$H(s) = \frac{V_2(s)}{V_1(s)} = \frac{Z_C(s)}{Z_L(s) + Z_R(s) + Z_C(s)} = \frac{\frac{1}{sC}}{sL + R + \frac{1}{sC}} = \frac{1}{s^2 LC + sRC + 1} = \frac{\frac{1}{LC}}{s^2 + s \frac{R}{L} + \frac{1}{LC}}$$

Comparando con la forma canónica $H(s) = \frac{\omega_0^2}{s^2 + \frac{\omega_0}{Q} s + \omega_0^2}$:
- $\omega_0 = \frac{1}{\sqrt{LC}} = 1\text{ rad/s}$
- $Q = \frac{\omega_0 L}{R} = \frac{1}{\sqrt{2}} \implies \mathbf{L = \frac{\sqrt{2}}{2}\text{ H} \approx 0.7071\text{ H}}$
- $C = \frac{1}{\omega_0^2 L} = \frac{1}{L} \implies \mathbf{C = \sqrt{2}\text{ F} \approx 1.4142\text{ F}}$
- Término lineal: $\frac{\omega_0}{Q} = \frac{1}{1/\sqrt{2}} = \mathbf{\sqrt{2} \approx 1.4142}$

La función de transferencia final es la respuesta de Butterworth de 2º orden:
$$\mathbf{H(s) = \frac{1}{s^2 + \sqrt{2} s + 1}}$$

**Respuesta de Frecuencia:**
Evaluando en $s = j\omega$:
$$H(j\omega) = \frac{1}{(1 - \omega^2) + j \sqrt{2}\omega}$$
$$|H(j\omega)| = \frac{1}{\sqrt{(1 - \omega^2)^2 + 2\omega^2}} = \frac{1}{\sqrt{1 - 2\omega^2 + \omega^4 + 2\omega^2}} = \mathbf{\frac{1}{\sqrt{1 + \omega^4}}}$$

- Para $\omega \ll 1$: $|H(j\omega)| \to 1 \implies \mathbf{0\text{ dB}}$, fase $\phi \to \mathbf{0^\circ}$.
- Para $\omega = \omega_0 = 1\text{ rad/s}$: $|H(j1)| = \frac{1}{\sqrt{2}} \implies \mathbf{-3.01\text{ dB}}$, fase $\phi(1) = \mathbf{-90^\circ}$.
- Para $\omega \gg 1$: $|H(j\omega)| \approx \frac{1}{\omega^2}$, con pendiente asintótica de $\mathbf{-40\text{ dB/década}}$ ($-12\text{ dB/octava}$), fase $\phi \to \mathbf{-180^\circ}$.

**Respuestas a las Preguntas de la Consigna:**
- *¿De qué tipo de filtro se trata?* Es un Filtro Pasa-Bajos (LPF) de segundo orden Butterworth. Tiene respuesta máximamente plana en la banda pasante (sin sobreelevación ni pico resonante) y atenúa las altas frecuencias a razón de $-40\text{ dB/década}$.
- *¿Cuál es el ancho de banda del filtro?* Como en $\omega = 1\text{ rad/s}$ la ganancia cae exactamente a $\frac{1}{\sqrt{2}}$ ($-3.01\text{ dB}$), la frecuencia de corte coincide con la frecuencia natural: $$\mathbf{B = \omega_c = \omega_0 = 1\text{ rad/s}} \quad \left(f_c = \frac{1}{2\pi} \approx 0.159\text{ Hz}\right)$$


```python
# Simulación y Gráfico de Bode — Circuito A (LPF Butterworth)
import numpy as np
import matplotlib.pyplot as plt

w = np.logspace(-2, 2, 1000)
s = 1j * w
alpha = np.sqrt(2)  # omega_0 / Q = sqrt(2)

num_A = 1.0
den = s**2 + alpha * s + 1.0
H_A = num_A / den
mag_A = 20 * np.log10(np.abs(H_A))
phase_A = np.angle(H_A, deg=True)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(9, 7), sharex=True)
fig.suptitle(r'Circuito A — Filtro Pasa-Bajos (LPF) Butterworth ($Q = \sqrt{2}/2$)', fontsize=12, fontweight='bold')

# Módulo
ax1.semilogx(w, mag_A, 'b-', linewidth=2, label=r'$|H(j\omega)|$')
ax1.plot([1.0], [-3.01], 'ro', markersize=6)
ax1.annotate(r'$\omega_c = \omega_0 = 1$ rad/s (-3.01 dB)', xy=(1, -3.01), xytext=(1.5, -12),
             arrowprops=dict(facecolor='crimson', shrink=0.08, width=1, headwidth=6), fontweight='bold', color='crimson')
ax1.semilogx([0.01, 1.0], [0, 0], 'k--', alpha=0.5, label='Asíntota 0 dB')
ax1.semilogx([1.0, 100.0], -40 * np.log10([1.0, 100.0]), 'g--', alpha=0.6, label='Asíntota -40 dB/década')
ax1.set_ylabel(r'Módulo $|H(j\omega)|$ [dB]')
ax1.set_ylim(-60, 10)
ax1.grid(True, which='both', alpha=0.4)
ax1.legend(loc='lower left', fontsize=9)

# Fase
ax2.semilogx(w, phase_A, 'royalblue', linewidth=2, label=r'$\phi(\omega)$')
ax2.plot([1.0], [-90], 'ro', markersize=6)
ax2.annotate(r'$\omega = 1$ rad/s (-90°)', xy=(1, -90), xytext=(1.5, -60),
             arrowprops=dict(facecolor='crimson', shrink=0.08, width=1, headwidth=6), fontweight='bold', color='crimson')
ax2.set_xlabel(r'Frecuencia angular $\omega$ [rad/s]')
ax2.set_ylabel(r'Fase $\phi(\omega)$ [grados]')
ax2.set_yticks([0, -45, -90, -135, -180])
ax2.set_ylim(-190, 10)
ax2.grid(True, which='both', alpha=0.4)
ax2.legend(loc='lower left', fontsize=9)

plt.tight_layout()
plt.savefig('circuito_a_bode.png')
plt.show()
```

### **Circuito B — Filtro Pasa-Altos (HPF)**

**Obtención Analítica de la Función de Transferencia $H(s)$:**
Trabajando en el dominio de Laplace en circuito abierto ($I_2 = 0$):
- Impedancia del bloque serie $R \parallel C$: $$Z_{RC}(s) = \frac{R \cdot \frac{1}{sC}}{R + \frac{1}{sC}} = \frac{R}{1 + sRC}$$
- Rama de salida (Inductor $L$): $Z_L(s) = sL$
- Divisor de Tensión:
$$H(s) = \frac{V_2(s)}{V_1(s)} = \frac{sL}{\frac{R}{1 + sRC} + sL} = \frac{sL(1 + sRC)}{R + sL(1 + sRC)} = \frac{s^2 RLC + sL}{s^2 RLC + sL + R} = \frac{s^2 + s \frac{1}{RC}}{s^2 + s \frac{1}{RC} + \frac{1}{LC}}$$

Comparando con la forma canónica $H(s) = \frac{s^2 + s \frac{\omega_0}{Q}}{s^2 + s \frac{\omega_0}{Q} + \omega_0^2}$:
- $\omega_0 = \frac{1}{\sqrt{LC}} = 1\text{ rad/s}$
- $\frac{\omega_0}{Q} = \frac{1}{RC} \implies Q = \omega_0 R C = \frac{1}{\sqrt{2}} \implies \mathbf{C = \frac{\sqrt{2}}{2}\text{ F} \approx 0.7071\text{ F}}$
- $L = \frac{1}{\omega_0^2 C} = \frac{1}{C} \implies \mathbf{L = \sqrt{2}\text{ H} \approx 1.4142\text{ H}}$
- Término lineal: $\frac{\omega_0}{Q} = \mathbf{\sqrt{2} \approx 1.4142}$

La función de transferencia analítica final es:
$$\mathbf{H(s) = \frac{s^2 + \sqrt{2} s}{s^2 + \sqrt{2} s + 1}}$$

**Respuesta de Frecuencia:**
Evaluando en $s = j\omega$:
$$H(j\omega) = \frac{-\omega^2 + j\sqrt{2}\omega}{(1 - \omega^2) + j\sqrt{2}\omega}$$
$$|H(j\omega)|^2 = \frac{(-\omega^2)^2 + 2\omega^2}{(1 - \omega^2)^2 + 2\omega^2} = \frac{\omega^4 + 2\omega^2}{\omega^4 + 1} \implies |H(j\omega)| = \mathbf{\sqrt{\frac{\omega^4 + 2\omega^2}{\omega^4 + 1}}}$$

- Para $\omega \ll 1$: $H(j\omega) \approx j\sqrt{2}\omega$. Sube con pendiente asintótica de $\mathbf{+20\text{ dB/década}}$, fase $\phi \to \mathbf{+90^\circ}$.
- Para $\omega = \omega_0 = 1\text{ rad/s}$:
  $$H(j1) = \frac{-1 + j\sqrt{2}}{j\sqrt{2}} = 1 + j\frac{1}{\sqrt{2}} \implies |H(j1)| = \sqrt{1 + \frac{1}{2}} = \sqrt{1.5} \approx 1.225 \quad (\mathbf{+1.76\text{ dB}})$$
  $$\phi(1) = \arctan\left(\frac{1}{\sqrt{2}}\right) \approx \mathbf{+35.26^\circ}$$
- Para $\omega \gg 1$: $|H(j\omega)| \to 1 \implies \mathbf{0\text{ dB}}$, fase $\phi \to \mathbf{0^\circ}$.

**Respuestas a las Preguntas de la Consigna:**
- *¿De qué tipo de filtro se trata?* Es un Filtro Pasa-Altos (HPF) de segundo orden. Posee un cero en el origen ($s=0$) que bloquea la continua y un cero real en $s = -\sqrt{2}$. Deja pasar las altas frecuencias sin atenuación ($0\text{ dB}$) y presenta una ligera sobreelevación de $+1.76\text{ dB}$ en $\omega = 1\text{ rad/s}$ provocada por el cero del numerador.
- *¿Cuál es el ancho de banda del filtro?* Se define a partir de la frecuencia de corte inferior $\omega_c$ donde la potencia cae a la mitad ($-3\text{ dB}$) respecto a la banda pasante ($|H|^2 = 1/2$):
$$\frac{\omega_c^4 + 2\omega_c^2}{\omega_c^4 + 1} = \frac{1}{2} \implies 2\omega_c^4 + 4\omega_c^2 = \omega_c^4 + 1 \implies \mathbf{\omega_c^4 + 4\omega_c^2 - 1 = 0}$$
Resolviendo para $\omega_c^2$:
$$\omega_c^2 = \frac{-4 + \sqrt{16 + 4}}{2} = \sqrt{5} - 2 \approx 0.2361 \implies \mathbf{\omega_c = \sqrt{\sqrt{5} - 2} \approx 0.486\text{ rad/s}} \quad \left(f_c \approx 0.077\text{ Hz}\right)$$
Por lo tanto, la banda pasante del filtro es $\mathbf{B = [\omega_c, \infty) \approx [0.486, \infty)\text{ rad/s}}$.


```python
# Simulación y Gráfico de Bode — Circuito B (HPF)
import numpy as np
import matplotlib.pyplot as plt

w = np.logspace(-2, 2, 1000)
s = 1j * w
alpha = np.sqrt(2)

num_B = s**2 + alpha * s
den = s**2 + alpha * s + 1.0
H_B = num_B / den
mag_B = 20 * np.log10(np.abs(H_B))
phase_B = np.angle(H_B, deg=True)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(9, 7), sharex=True)
fig.suptitle(r'Circuito B — Filtro Pasa-Altos (HPF) ($Q = \sqrt{2}/2$)', fontsize=12, fontweight='bold')

# Módulo
ax1.semilogx(w, mag_B, 'b-', linewidth=2, label=r'$|H(j\omega)|$')
w_c_B = float(np.sqrt(np.sqrt(5) - 2))
ax1.plot([w_c_B], [-3.01], 'ro', markersize=6)
ax1.annotate(r'$\omega_c \approx 0.49$ rad/s (-3 dB)', xy=(w_c_B, -3.01), xytext=(0.03, -12),
             arrowprops=dict(facecolor='crimson', shrink=0.08, width=1, headwidth=6), fontweight='bold', color='crimson')
ax1.plot([1.0], [1.76], 'mo', markersize=6)
ax1.annotate(r'$\omega_0 = 1$ rad/s (+1.76 dB)', xy=(1, 1.76), xytext=(1.8, 6),
             arrowprops=dict(facecolor='purple', shrink=0.08, width=1, headwidth=6), fontweight='bold', color='purple')
ax1.semilogx([0.01, 1.0], 20 * np.log10(alpha * np.array([0.01, 1.0])), 'g--', alpha=0.6, label='Asíntota +20 dB/década')
ax1.semilogx([1.0, 100.0], [0, 0], 'k--', alpha=0.5, label='Asíntota 0 dB')
ax1.set_ylabel(r'Módulo $|H(j\omega)|$ [dB]')
ax1.set_ylim(-50, 15)
ax1.grid(True, which='both', alpha=0.4)
ax1.legend(loc='lower right', fontsize=9)

# Fase
ax2.semilogx(w, phase_B, 'royalblue', linewidth=2, label=r'$\phi(\omega)$')
ax2.plot([1.0], [35.26], 'ro', markersize=6)
ax2.annotate(r'$\omega = 1$ rad/s (+35.3°)', xy=(1, 35.26), xytext=(1.5, 60),
             arrowprops=dict(facecolor='crimson', shrink=0.08, width=1, headwidth=6), fontweight='bold', color='crimson')
ax2.set_xlabel(r'Frecuencia angular $\omega$ [rad/s]')
ax2.set_ylabel(r'Fase $\phi(\omega)$ [grados]')
ax2.set_yticks([0, 30, 60, 90])
ax2.set_ylim(-10, 100)
ax2.grid(True, which='both', alpha=0.4)
ax2.legend(loc='lower left', fontsize=9)

plt.tight_layout()
plt.savefig('circuito_b_bode.png')
plt.show()
```

### **Circuito C — Filtro Pasa-Banda (BPF)**

**Obtención Analítica de la Función de Transferencia $H(s)$:**
Trabajando en el dominio de Laplace en circuito abierto ($I_2 = 0$):
- Impedancia del tanque paralelo $L \parallel C$: $$Z_p(s) = \frac{sL \cdot \frac{1}{sC}}{sL + \frac{1}{sC}} = \frac{sL}{s^2 LC + 1}$$
- Divisor de Tensión con la resistencia serie $R$:
$$H(s) = \frac{V_2(s)}{V_1(s)} = \frac{Z_p(s)}{R + Z_p(s)} = \frac{sL}{s^2 RLC + sL + R} = \frac{s \frac{1}{RC}}{s^2 + s \frac{1}{RC} + \frac{1}{LC}}$$

Comparando con la forma canónica $H(s) = \frac{s \frac{\omega_0}{Q}}{s^2 + s \frac{\omega_0}{Q} + \omega_0^2}$:
- $\omega_0 = \frac{1}{\sqrt{LC}} = 1\text{ rad/s}$
- $\frac{\omega_0}{Q} = \frac{1}{RC} \implies Q = \omega_0 R C = \frac{1}{\sqrt{2}} \implies \mathbf{C = \frac{\sqrt{2}}{2}\text{ F} \approx 0.7071\text{ F}}$
- $L = \frac{1}{\omega_0^2 C} = \frac{1}{C} \implies \mathbf{L = \sqrt{2}\text{ H} \approx 1.4142\text{ H}}$
- Término del numerador y lineal: $\frac{\omega_0}{Q} = \mathbf{\sqrt{2} \approx 1.4142}$

La función de transferencia analítica es:
$$\mathbf{H(s) = \frac{\sqrt{2} s}{s^2 + \sqrt{2} s + 1}}$$

**Respuesta de Frecuencia:**
Evaluando en $s = j\omega$:
$$H(j\omega) = \frac{j \sqrt{2}\omega}{(1 - \omega^2) + j \sqrt{2}\omega}$$
$$|H(j\omega)| = \frac{\sqrt{2}\omega}{\sqrt{(1 - \omega^2)^2 + 2\omega^2}} = \mathbf{\frac{\sqrt{2}\omega}{\sqrt{1 + \omega^4}}}$$

- Para $\omega \ll 1$: $H(j\omega) \approx j\sqrt{2}\omega$, pendiente de $\mathbf{+20\text{ dB/década}}$, fase $\phi \to \mathbf{+90^\circ}$.
- Para $\omega = \omega_0 = 1\text{ rad/s}$:
  $$H(j1) = \frac{j\sqrt{2}}{j\sqrt{2}} = 1 \implies |H(j1)| = 1 \quad (\mathbf{0\text{ dB}}), \quad \phi(1) = \mathbf{0^\circ}$$
- Para $\omega \gg 1$: $H(j\omega) \approx \frac{\sqrt{2}}{j\omega}$, pendiente de $\mathbf{-20\text{ dB/década}}$, fase $\phi \to \mathbf{-90^\circ}$.

**Respuestas a las Preguntas de la Consigna:**
- *¿De qué tipo de filtro se trata?* Es un Filtro Pasa-Banda (BPF) de segundo orden. Posee un cero en el origen ($s=0$) y transmite con máxima ganancia ($0\text{ dB}$) únicamente en la banda centrada en $\omega_0 = 1\text{ rad/s}$.
- *¿Cuál es el ancho de banda del filtro?* En un filtro pasa-banda canónico de segundo orden, el ancho de banda $B$ a $-3\text{ dB}$ está dado directamente por la relación:
$$\mathbf{B = \frac{\omega_0}{Q} = \frac{1}{1/\sqrt{2}} = \sqrt{2} \approx 1.4142\text{ rad/s}} \quad \left(f_B = \frac{\sqrt{2}}{2\pi} \approx 0.225\text{ Hz}\right)$$
Las frecuencias de corte a $-3\text{ dB}$ se obtienen igualando $|H|^2 = 1/2$:
$$\frac{2\omega^2}{1 + \omega^4} = \frac{1}{2} \implies \omega^4 - 4\omega^2 + 1 = 0 \implies \omega^2 = 2 \pm \sqrt{3}$$
$$\mathbf{\omega_{c1} = \sqrt{2 - \sqrt{3}} \approx 0.518\text{ rad/s}}, \quad \mathbf{\omega_{c2} = \sqrt{2 + \sqrt{3}} \approx 1.932\text{ rad/s}}$$
Verificación: $B = \omega_{c2} - \omega_{c1} = \sqrt{2+\sqrt{3}} - \sqrt{2-\sqrt{3}} = \sqrt{2} \approx 1.4142\text{ rad/s}$.


```python
# Simulación y Gráfico de Bode — Circuito C (BPF)
import numpy as np
import matplotlib.pyplot as plt

w = np.logspace(-2, 2, 1000)
s = 1j * w
alpha = np.sqrt(2)

num_C = alpha * s
den = s**2 + alpha * s + 1.0
H_C = num_C / den
mag_C = 20 * np.log10(np.abs(H_C))
phase_C = np.angle(H_C, deg=True)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(9, 7), sharex=True)
fig.suptitle(r'Circuito C — Filtro Pasa-Banda (BPF) ($Q = \sqrt{2}/2$)', fontsize=12, fontweight='bold')

# Módulo
ax1.semilogx(w, mag_C, 'b-', linewidth=2, label=r'$|H(j\omega)|$')
ax1.plot([1.0], [0.0], 'mo', markersize=6)
ax1.annotate(r'$\omega_0 = 1$ rad/s (0 dB)', xy=(1, 0), xytext=(1.5, 5),
             arrowprops=dict(facecolor='purple', shrink=0.08, width=1, headwidth=6), fontweight='bold', color='purple')
w_c1 = float(np.sqrt(2 - np.sqrt(3)))
w_c2 = float(np.sqrt(2 + np.sqrt(3)))
ax1.plot([w_c1, w_c2], [-3.01, -3.01], 'ro', markersize=6)
ax1.annotate(r'$\omega_{c1} \approx 0.52$ rad/s', xy=(w_c1, -3.01), xytext=(0.04, -15),
             arrowprops=dict(facecolor='crimson', shrink=0.08, width=1, headwidth=6), fontweight='bold', color='crimson')
ax1.annotate(r'$\omega_{c2} \approx 1.93$ rad/s', xy=(w_c2, -3.01), xytext=(3, -15),
             arrowprops=dict(facecolor='crimson', shrink=0.08, width=1, headwidth=6), fontweight='bold', color='crimson')
ax1.set_ylabel(r'Módulo $|H(j\omega)|$ [dB]')
ax1.set_ylim(-50, 15)
ax1.grid(True, which='both', alpha=0.4)
ax1.legend(loc='lower left', fontsize=9)

# Fase
ax2.semilogx(w, phase_C, 'royalblue', linewidth=2, label=r'$\phi(\omega)$')
ax2.plot([1.0], [0.0], 'ro', markersize=6)
ax2.annotate(r'$\omega = 1$ rad/s (0°)', xy=(1, 0), xytext=(1.5, 30),
             arrowprops=dict(facecolor='crimson', shrink=0.08, width=1, headwidth=6), fontweight='bold', color='crimson')
ax2.set_xlabel(r'Frecuencia angular $\omega$ [rad/s]')
ax2.set_ylabel(r'Fase $\phi(\omega)$ [grados]')
ax2.set_yticks([-90, -45, 0, 45, 90])
ax2.set_ylim(-100, 100)
ax2.grid(True, which='both', alpha=0.4)
ax2.legend(loc='lower left', fontsize=9)

plt.tight_layout()
plt.savefig('circuito_c_bode.png')
plt.show()
```

## **Análisis de Resultados**

**Polos en Común y Estabilidad:**
Al fijar $\omega_0 = 1\text{ rad/s}$ y $Q = \frac{1}{\sqrt{2}} \approx 0.7071$, las tres funciones de transferencia comparten exactamente el mismo denominador:
$$D(s) = s^2 + \sqrt{2} s + 1 = 0$$
cuyos polos son:
$$s_{1,2} = -\frac{\sqrt{2}}{2} \pm j \frac{\sqrt{2}}{2} \approx -0.7071 \pm j 0.7071$$
- *Ubicación:* Los polos se encuentran en el semiplano izquierdo sobre la circunferencia unitaria ($|s| = 1$), formando un ángulo de $\pm 135^\circ$ con el eje real positivo. Esto garantiza que los tres circuitos sean estrictamente BIBO estables.
- *Dinámica temporal:* Al tener parte real $\sigma = -0.7071$, los tres presentan el mismo factor de amortiguamiento exponencial transitorio $\zeta = \frac{1}{2Q} = \frac{\sqrt{2}}{2} \approx 0.7071$.

**El Rol de los Ceros:**
Si los polos son idénticos, la función del filtro queda gobernada exclusivamente por los ceros del numerador (es decir, el componente en el que se mide la salida $V_2$):
- *Circuito A (Salida en $C$):* No tiene ceros finitos. En $\omega = 0$ transmite todo ($0\text{ dB}$); a altas frecuencias el capacitor se cortocircuita derivando la señal a masa ($-40\text{ dB/década}$). Actúa como Pasa-Bajos.
- *Circuito B (Salida en $L$ con $R \parallel C$ en serie):* Tiene un cero en el origen ($s=0$) y un cero real en $s = -\sqrt{2}$. En baja frecuencia el inductor cortocircuita la salida a masa ($+20\text{ dB/década}$); a alta frecuencia el capacitor serie actúa como cortocircuito transfiriendo toda la señal al inductor ($0\text{ dB}$). Actúa como Pasa-Altos.
- *Circuito C (Salida en $L \parallel C$ con $R$ en serie):* Tiene un cero en el origen ($s=0$). En bajas frecuencias el inductor deriva a masa y en altas frecuencias el capacitor deriva a masa. Solo en la resonancia paralela (antirresonancia) de $L \parallel C$ en $\omega = 1\text{ rad/s}$ la impedancia es infinita y no carga a $R$, logrando transmisión máxima ($0\text{ dB}$). Actúa como Pasa-Banda.

**Significado de $Q = \frac{\sqrt{2}}{2} \approx 0.7071$ (Butterworth):**
A diferencia de un sistema fuertemente subamortiguado con $Q > 1/\sqrt{2}$ (donde aparecen picos pronunciados de sobreelevación), el valor $Q = 1/\sqrt{2}$ representa el punto de máxima planicidad (Butterworth):
- En el Pasa-Bajos no existe sobreimpulso resonante; el módulo decae suave y monótonamente, y la frecuencia de corte a $-3\text{ dB}$ cae exactamente en la frecuencia natural $\omega_c = \omega_0 = 1\text{ rad/s}$.
- En el Pasa-Banda, la selectividad se ensancha: el ancho de banda es exactamente $B = \frac{\omega_0}{Q} = \sqrt{2} \approx 1.414\text{ rad/s}$, cubriendo una banda pasante más amplia que en diseños de alto $Q$.

## **Aprendizajes**

- *Diseño de Butterworth ($Q = 1/\sqrt{2}$):* Comprobamos que al fijar $Q = 1/\sqrt{2} \approx 0.7071$, el filtro Pasa-Bajos adopta la respuesta clásicamente conocida como Butterworth de segundo orden, donde la ganancia es máximamente plana en la banda de paso y la atenuación a $-3\text{ dB}$ ocurre con precisión en $\omega_0 = 1\text{ rad/s}$.

- *Los ceros determinan la selectividad de banda:* Confirmamos que aunque la malla RLC impone los mismos polos en el denominador (garantizando estabilidad y la misma respuesta transitoria), la elección del puerto de salida $V_2$ define los ceros del sistema y transforma la naturaleza del filtro entre Pasa-Bajos, Pasa-Altos y Pasa-Banda.

- *Relación entre topología y componentes:* Vimos que para mantener $\omega_0 = 1\text{ rad/s}$ y $Q = \sqrt{2}/2$, la ubicación de los componentes reactivos permuta sus valores: en el pasa-bajos serie $L = \sqrt{2}/2\text{ H}$ y $C = \sqrt{2}\text{ F}$, mientras que al interactuar en paralelo con la resistencia se invierten los roles a $C = \sqrt{2}/2\text{ F}$ y $L = \sqrt{2}\text{ H}$.

- *Comportamiento de la fase en resonancia:* En $\omega_0 = 1\text{ rad/s}$, el filtro Pasa-Banda tiene desfase de $0^\circ$ (comportamiento puramente resistivo por antirresonancia de $L \parallel C$), mientras que el Pasa-Bajos se encuentra a $-90^\circ$ y el Pasa-Altos a $+35.26^\circ$.
